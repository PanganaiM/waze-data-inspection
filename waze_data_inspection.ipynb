## 4. Data Loading & Initial Inspection

```python
# Import packages for data manipulation
import pandas as pd
import numpy as np
```

```python
# Load dataset into dataframe
df = pd.read_csv('waze_dataset.csv')
```

### Initial data-quality findings

The dataset contains **14,999 rows and 13 columns**. There are **700 missing values**, all in the churn `label` field. The initial inspection provides the basis for determining whether the missing labels appear systematically different from the labelled observations.

## 5. Missing-Label Assessment

Summary statistics are compared between the 700 observations with missing labels and the observations with known labels.

```python
# Isolate rows with null values
df_null = df[df['label'].isna()]

# Display summary stats of rows with null values
df_null.describe()
```

```python
# Isolate rows without null values
df_not_null = df[df['label'].notna()]

# Display summary stats of rows without null values
df_not_null.describe()
```

The two groups are broadly similar across their summary statistics, providing no obvious evidence from this comparison that the missing labels follow a strong systematic behavioural pattern.

## 6. Missing Labels by Device

```python
# Get count of null values by device
df_null['device'].value_counts()
```

```python
# Calculate % of iPhone nulls and Android nulls
df_null['device'].value_counts(normalize=True) * 100
```

```python
# Calculate % of iPhone users and Android users in full dataset
df['device'].value_counts(normalize=True) * 100
```

Among missing-label observations, **447 are iPhone users** and **253 are Android users**. This is about **63.86% iPhone / 36.14% Android**, close to the full dataset distribution of **64.48% iPhone / 35.52% Android**. Device representation therefore does not suggest an obvious device-related missingness pattern.

## 7. Churn Distribution

```python
# Calculate counts of churned vs. retained
df['label'].value_counts()

```

Among users with known labels, **11,763 are retained** and **2,536 churned**, corresponding to approximately **82% retained and 18% churned**.

## 8. Retained vs Churned Behaviour

Median statistics are used to reduce the influence of extreme values when comparing the two groups.

```python
# Calculate median values of all columns for churned and retained users
df.groupby('label').median(numeric_only=True)
```

The comparison shows that churned users have slightly more recent drives, but retained users are active on substantially more days. This suggests that engagement frequency may be more informative than raw drive counts alone.

### Distance per drive

```python
# Add a column to df called `km_per_drive`
df['km_per_drive'] = df['driven_km_drives'] / df['drives']

# Group by `label`, calculate the median, and isolate for km per drive
df.groupby('label')['km_per_drive'].median()
```

Median kilometres per drive are very similar: approximately **74.11 km for churned users** and **75.01 km for retained users**.

### Distance per driving day

```python
# Add a column to df called `km_per_driving_day`
df['km_per_driving_day'] = df['driven_km_drives'] / df['driving_days']

# Group by `label`, calculate the median, and isolate for km per driving day
df.groupby('label')['km_per_driving_day'].median()
```

Median kilometres per driving day are much higher for churned users (**≈697.54 km**) than retained users (**≈289.55 km**). The magnitude warrants careful validation and later outlier analysis.

### Drives per driving day

```python
# Add a column to df called `drives_per_driving_day`
df['drives_per_driving_day'] = df['drives'] / df['driving_days']

# Group by `label`, calculate the median, and isolate for drives per driving day
df.groupby('label')['drives_per_driving_day'].median()
```

The median churned user records **10 drives per driving day**, compared with about **4.06** for retained users. Combined with fewer active days, this suggests a concentrated-use pattern among churned users.

## 9. Churn by Device

```python
# For each label, calculate the number of Android users and iPhone users
# For each label, calculate the number of Android users and iPhone users
df.groupby('label')['device'].value_counts()
```

```python
# For each label, calculate the percentage of Android users and iPhone users
# For each label, calculate the percentage of Android users and iPhone users
df.groupby('label')['device'].value_counts(normalize=True) * 100
```

The device mix is almost identical across churned and retained users: roughly **65% iPhone and 35% Android** in both groups. Device type therefore does not appear to explain the churn split at this inspection stage.

## 10. Key Findings

- The dataset contains 14,999 users and 13 variables.
- 700 churn labels are missing; the inspected summaries do not show an obvious systematic pattern.
- Known-label users are approximately 82% retained and 18% churned.
- Device proportions are highly consistent across missing-label, churned and retained groups.
- Retained users are active on more days, while churned users show more concentrated driving behaviour on the days they drive.
- Very high kilometres-per-driving-day values require further validation and outlier analysis.
